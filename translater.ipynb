```python
!pip install -q transformers sentencepiece gradio

import torch
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load a multilingual translation model
MODEL_NAME = "facebook/nllb-200-distilled-600M"

print("Loading English-to-Tamil model...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Model loaded successfully!")
print("Device:", device)


# English to Tamil translation
def translate(text):
    if not text or not text.strip():
        return "Please enter English text."

    inputs = tokenizer(
        text.strip(),
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    # Set the target language to Tamil
    tamil_token_id = tokenizer.convert_tokens_to_ids("tam_Taml")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            forced_bos_token_id=tamil_token_id,
            max_new_tokens=256,
            num_beams=4
        )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )


# Create the translator interface
demo = gr.Interface(
    fn=translate,
    inputs=gr.Textbox(
        lines=5,
        label="English Text",
        placeholder="Enter English text here..."
    ),
    outputs=gr.Textbox(
        lines=5,
        label="Tamil Translation"
    ),
    title="English to Tamil Translator",
    description="Translate English sentences and paragraphs into Tamil.",
    examples=[
        ["Hello, how are you?"],
        ["I am learning Python."],
        ["India is a beautiful country."],
        ["Education is important for everyone."]
    ]
)

demo.launch(share=True)
```
